```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
```

```python
plt.plot(x, np.sin(x), '--')
plt.plot(x, np.cos(x), '-')
```

```python
fig = plt.figure()
plt.plot(x, np.log(x+1))
plt.plot(x, np.cos(x))
fig.savefig('my_matfig.png')
```
