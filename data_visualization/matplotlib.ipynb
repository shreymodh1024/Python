```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,10)
y = x**2
type(y)
```

```python
fig1 = plt.figure()
axis1 = fig1.add_axes([0.1, 0.1, 1,1])
axis1.plot(x,y)
axis2 = fig1.add_axes([0.2,0.5, .5,.5])
axis2.plot(x,y)
```

```python
fig2 = plt.figure(figsize=(3,3))
axis3 = fig2.add_axes([0.1,0.1,.4,.8])
axis4 = fig2.add_axes([.3,.4,.3,.3])
```

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig('basicplot.png')
```
